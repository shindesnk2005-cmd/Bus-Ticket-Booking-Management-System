```python
#convert data columns
date_columns =[
    "Journey_Date",
    "Booking_Date",
    "Cancellation_Date"
]
```

```python
#nuberical data conversion
# Numerical Data Conversion

numeric_columns = [
    "Age",
    "Distance_KM",
    "Ticket_Price",
    "Occupancy_Percent"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("Numerical Conversion Completed!")
```

```python
#clean text columns
text_columns =[
    "Gender",
    "Bus_Type",
    "Source",
    "Destination",
    "Payment_Status",
    "Payment_Status",
    "Booking_Status"
    ]
for col in text_columns:
    df[col] = df[col].astype("string").str.strip()
    print("Text Cleaning Completed!")
```

```python
#check invalid values
print("Invalid Age:", ((df["Age"] <1) | (df["Age"] >100)).sum())

print("Invalid Distance:", (df["Distance_KM"] <=0).sum())
print("Invalid Ticket Price:",(df["Ticket_Price"] <=0).sum())
        
print("Invalid Occupancy:",(df["Occupancy_Percent"] <0)| (df["Occupancy_Percent"] > 100).sum())      
```

```python
#final missing value check
df.isnull().sum()
```

```python
#find dataset check
print("Final Rows:", df.shape[0])
print("Final Columns:", df.shape[1])
print("Duplicate Records:",df.duplicated().sum())
```

```python
#data transformation
#create year column
df["Journey_Year"] = df["Journey_Date"].dt.year
```

```python
#create month number
df["Month_Number"] = df["Journey_Date"].dt.month
```

```python
#create month name
df["Month_Name"] = df["Journey_Date"].dt.month_name()
```

```python
#create day name
df["Month_Name"] = df["Journey_Date"].dt.month_name()
```

```python
#create route column
df["Route"] = df["Source"] + " - " + df["Destination"]
```

```python
#create passenger age group
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[0, 18, 30, 45, 60, 100],
    labels=["Below 18", "18-30", "31-45", "46-60", "Above 60"]
)
```

```python
#create booking lesd days
df["Booking_Lead_Days"]=(df["Journey_Date"]-df["Booking_Date"]).dt.days
```

```python
#display update datset
df.head()
```

```python
#check new columns
print(df.columns)
```

```python
#create analysis columns
df["Booking_Count"]=1
```

```python
#revenue calculation
payment = df["Payment_Status"].str.lower()
booking = df["Booking_Status"].str.lower()

df["Revenue_Amount"] = np.where(
    (payment == "paid") & (booking == "confirmed"),
    df["Ticket_Price"].fillna(0),
    0
)

print("Revenue Column Created!")
```

```python
#cancellation flag
df["Cancellation_Flag"] = np.where(
    booking == "cancelled",
    1,
    0
)
```

```python
#booking status bar chart
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Booking_Status")

plt.title("Booking Status Analysis")
plt.xlabel("Booking Status")
plt.ylabel("Total Bookings")
plt.xticks(rotation=30)

plt.tight_layout()
plt.savefig("Graphs/booking_status.png")
plt.show()
```

```python
#bustype analysis
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="Bus_Type")

plt.title("Bus Type Analysis")
plt.xlabel("Bus Type")
plt.ylabel("Total Bookings")

plt.tight_layout()
plt.savefig("Graphs/bus_type.png")
plt.show()
```

```python
#payment method pie chart
payment_counts = df["Payment_Method"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    payment_counts,
    labels=payment_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Payment Method Analysis")
plt.savefig("Graphs/payment_method.png")
plt.show()
```

```python
#monthly booking line chart
monthly = (
    df.groupby(df["Journey_Date"].dt.to_period("M"))
    .size()
)

plt.figure(figsize=(12, 5))

plt.plot(monthly.index.astype(str), monthly.values, marker="o")

plt.title("Monthly Booking Trend")
plt.xlabel("Month")
plt.ylabel("Total Bookings")
plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig("Graphs/monthly_bookings.png")
plt.show()
```

```python
#reenue by bus type
revenue_bus = df.groupby("Bus_Type")["Revenue_Amount"].sum()

plt.figure(figsize=(8, 5))

revenue_bus.plot(kind="bar")

plt.title("Revenue by Bus Type")
plt.xlabel("Bus Type")
plt.ylabel("Revenue")

plt.tight_layout()
plt.savefig("Graphs/revenue_by_bus.png")
plt.show()
```

```python
#top 10 popular routes
top_routes = df["Route"].value_counts().head(10)

plt.figure(figsize=(12, 6))

top_routes.sort_values().plot(kind="barh")

plt.title("Top 10 Popular Routes")
plt.xlabel("Total Bookings")
plt.ylabel("Route")

plt.tight_layout()
plt.savefig("Graphs/top_routes.png")
plt.show()
```

```python
#passenger age distribution
plt.figure(figsize=(9, 5))

sns.histplot(data=df, x="Age", bins=15, kde=True)

plt.title("Passenger Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.tight_layout()
plt.savefig("Graphs/age_distribution.png")
plt.show()
```

```python
#age group analysis
plt.figure(figsize=(9, 5))

sns.countplot(data=df, x="Age_Group")

plt.title("Passenger Age Group Analysis")
plt.xlabel("Age Group")
plt.ylabel("Total Passengers")

plt.tight_layout()
plt.savefig("Graphs/age_group.png")
plt.show()
```

```python
#distance vs ticket price
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df,
    x="Distance_KM",
    y="Ticket_Price",
    hue="Bus_Type"
)

plt.title("Distance vs Ticket Price")
plt.xlabel("Distance in KM")
plt.ylabel("Ticket Price")

plt.tight_layout()
plt.savefig("Graphs/distance_price.png")
plt.show()
```

```python
#correlation heatmap
numeric_data = df[
    ["Age", "Distance_KM", "Ticket_Price", "Occupancy_Percent"]
]

plt.figure(figsize=(8, 6))

sns.heatmap(
    numeric_data.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.savefig("Graphs/correlation_heatmap.png")
plt.show()
```
